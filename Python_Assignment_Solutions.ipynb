## Q3.3

**Question:** Create `x = 100`, `y = 100`, and check `x is y` and `x == y`. Then create `p = 1000`, `q = 1000` and check the same. Discuss why small integers may behave differently from large ones.

**Answer:** Python commonly caches/reuses small integer objects, so two variables containing a small integer may refer to the same object and make `is` appear True. Larger integers are not guaranteed to be shared, so `is` can be False even when `==` is True. Use `==` for value comparison and `is` mainly for identity checks such as `x is None`.

```python
x = 100
y = 100
print("100:", x is y, x == y)

p = 1000
q = 1000
print("1000:", p is q, p == q)
```

## Q7.1

**Question:** Write `my_abs(val)` that returns the absolute value without using Python's built-in `abs()`. Test positive, negative, and 0.

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```

## Q7.2

**Question:** What will `my_abs("Hi")` do? Predict the error, then run it to confirm.

**Answer:** It raises a `TypeError` because the function tries to evaluate `"Hi" < 0`, and Python cannot compare a string with an integer.

```python
try:
    print(my_abs("Hi"))
except TypeError as e:
    print(type(e).__name__ + ":", e)
```

## Q7.3

**Question:** Write `print_abs(val)` that prints (not returns) the absolute value. Then run `x = print_abs(-2.7); print(x)`. What gets printed for x, and why?

**Answer:** The function prints `2.7`, but `x` becomes `None` because the function has no `return` statement. Therefore the second print displays `None`.

```python
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

## Q7.4

**Question:** Predict the output and explain why reassigning `val` inside the function does not change `x`.

**Answer:** Output: `7`. The function's parameter `val` is a local name. Reassigning it to `val + 1` only changes that local name; it does not reassign the caller's variable `x`.

```python
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)
```

## Q8.3

**Question:** Write `increment_counter()` that tries `counter = counter + 1` without `global`. Run it and explain the error.

**Answer:** Python treats `counter` as a local variable because the function assigns to it. When evaluating `counter + 1`, that local variable has not been assigned yet, so Python raises `UnboundLocalError`.

```python
counter = 0

def increment_counter():
    try:
        counter = counter + 1
    except UnboundLocalError as e:
        print(type(e).__name__ + ":", e)

increment_counter()
```

## Q8.4

**Question:** Fix the previous function using the `global` keyword so it correctly increments the global counter.

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
