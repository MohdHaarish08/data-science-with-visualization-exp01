### 29. Program covering the requested Series operations.

```python
sales = Series([450, 650, 300, 800], index=['Pen', 'Bag', 'Book', 'Laptop'])

print('Values:', sales.values)
print('Index:', sales.index)
print('Bag sales:', sales['Bag'])
print('Sales above 500:')
print(sales[sales > 500])
print('Does Pen exist?', 'Pen' in sales.index)
sales_dict = sales.to_dict()
print('Dictionary:', sales_dict)
```

### 30. Analyse the complete code.

```python
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'

print('(a) Index order:', list(s.index))
print('(b) Missing label: D')
print('(c) Values greater than 20:')
print(s[s > 20])
print('(d) Non-missing values:')
print(pd.notnull(s))
print('(e) Series name:', s.name)
print('    Index name:', s.index.name)
s
```
